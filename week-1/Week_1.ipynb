## 2. Loading and cleaning functions

`cleanup` loads the CSV and cleans it:
- strips extra spaces from names
- converts Age and Score to numbers (`errors="coerce"` turns invalid values like "abc" into NaN)
- drops rows with missing values
- keeps only ages 5-100 and scores 0-100
- removes exact duplicates

```python
def cleanup(path):
    try:
        df = pd.read_csv(path)
    except Exception as e:
        print("Error during file importing: ", str(e))
        return

    df["Name"] = df["Name"].str.strip()

    df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
    df["Score"] = pd.to_numeric(df["Score"], errors="coerce")

    df = df.dropna(subset=["Name", "Age", "Score"])

    df = df[df["Age"].between(5, 100) & df["Score"].between(0, 100)]

    df = df.drop_duplicates()

    df["Age"] = df["Age"].astype(int)

    return df.reset_index(drop=True)
```

## 3. NumPy statistics functions

The scores are converted to a NumPy array, then each function calculates one statistic. The pass rate works because `scores >= 50` gives True/False values, and the mean of those is the fraction that passed.

```python
def get_scores(df):
    return df["Score"].to_numpy(dtype=float)

def average_score(scores):
    return np.mean(scores)

def highest_score(scores):
    return np.max(scores)

def lowest_score(scores):
    return np.min(scores)

def total_score(scores):
    return np.sum(scores)

def pass_rate(scores, pass_mark=50):
    return np.mean(scores >= pass_mark) * 100
```

## 5. Cleaned data

```python
df = cleanup("students.csv")
print("Cleaned rows:", len(df))
df
```

## 6. Final report

```python
if df is None or df.empty:
    print("No valid records to process.")
else:
    scores = get_scores(df)

    print("===== STUDENT SCORE REPORT =====")
    print("Number of valid students:", len(scores))
    print("Average score:", average_score(scores))
    print("Highest score:", highest_score(scores))
    print("Lowest score:", lowest_score(scores))
    print("Total score:", total_score(scores))
    print(f"Pass rate: {pass_rate(scores):.1f}%")
```
